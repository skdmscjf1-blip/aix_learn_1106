```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn import svm,metrics
```

```python
# 1. 파일가져오기
# 2. 데이터 전처리
# 3. 학습/훈련데이터 분리
# 4. 모델 학습
# 5. 모델 평가
# 6. 모델 예측(확률)
```

```python
# 데이터 전처리
# 데이터 학습
# 데이터 예측
# 데이터 검증(정답률)
```

```python
df = pd.read_csv('data/iris.csv',index_col='caseno')
df.head()
```

```python
# 데이터 전처리 - 문자가 포함되어 있으면 학습이 안됨. 
df.columns
```

```python
df['Species'].unique()
```

```python
# 데이터 전처리 - 데이터 분리
data = df[['SepalLength','SepalWidth','PetalLength','PetalWidth']]
data.head() # 질문 - 문제 
```

```python
label = df['Species']
label.head(2)
```

```python
def nameChange(x):
    if x == 'Iris-setosa':
        return 0
    elif x == 'Iris-versicolor':
        return 1
    else :
        return 2
```

```python
label = label.apply(nameChange)
```

```python
label.unique()
```

```python
data.head()
label.head()
```

```python
# 학습데이터 / 테스트 데이터 분리 default =  0.25 , random_state = 42
train_data,test_data,train_label,test_label = \
    train_test_split(data,label,stratify=label,test_size=0.3,random_state=42)
```

```python
# 모델 학습
clf = svm.SVC()
clf.fit(train_data,train_label)
```

```python
# 모델 예측
result = clf.predict(test_data)
result
```

```python
score = metrics.accuracy_score(result,test_label)
print('정답률 : ',score)
```
