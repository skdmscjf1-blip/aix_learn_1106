```python
import joblib
import os
from pathlib import Path
import numpy as np
from scipy.special import softmax
```

```python
model_ex = joblib.load('model_ex.joblib')
```

```python
# 290.0,26.3,31.2,12.48,4.3056
in_data = input('데이터 입력 : 290.0,26.3,31.2,12.48,4.3056 ')
in_data = in_data.split(',')
in_data = list(map(float,in_data))
print('입력데이터 : ',in_data)
```

```python
print(model_ex.classes_)
```

```python
result = model_ex.predict([in_data])
result2 = model_ex.predict_proba([in_data])
decisions = model_ex.decision_function([in_data])
decisions
soft1 = softmax(decisions,axis=1)
result3 = np.round(soft1,decimals=3)[0]
if result[0] == 'Bream':
    result4 = result3[0]
elif result[0] == 'Parkki':
    result4 = result3[1]
elif result[0] == 'Perch':
    result4 = result3[2]
elif result[0] == 'Pike':
    result4 = result3[3]
elif result[0] == 'Roach':
    result4 = result3[4]
elif result[0] == 'Smelt':
    result4 = result3[5]
elif result[0] == 'Whitefish':
    result4 = result3[6]

print(f'예측생선 : {result[0]} (예측확률 : {result4})')
print('전체확률 : ',result3)

```
