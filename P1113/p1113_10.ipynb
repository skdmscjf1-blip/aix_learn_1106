```python
#그래프관련 import
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
# matplotlib.rcParams['font.family'] = 'AppleGothic'
matplotlib.rcParams['axes.unicode_minus']=False

import pandas as pd
import numpy as np
from sklearn import svm,metrics
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier,KNeighborsRegressor
from sklearn.linear_model import LinearRegression,Ridge,Lasso,LogisticRegression
from sklearn.preprocessing import StandardScaler,PolynomialFeatures,OneHotEncoder
from sklearn.metrics import mean_absolute_error
from scipy.special import expit,softmax # expit : 이진분류함수 , softmax : 다중분류 0-1 숫자로 변환
from sklearn.linear_model import SGDClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import plot_tree
from sklearn.model_selection import cross_validate
from sklearn.model_selection import StratifiedKFold
from sklearn.model_selection import GridSearchCV
```

```python
# 데이터 가져오기
wine = pd.read_csv('data/wine.csv')
wine.head(2)
data = wine[wine.columns[:-1]].values
label = wine['class'].values
# 데이터 분리
train_data,test_data,train_label,test_label = train_test_split(data,label,test_size=0.2,random_state=42)
```

```python
dt = DecisionTreeClassifier(random_state=42)
# params = {'min_inpurity_decrease':[0.0001,0.0002,0.0003,0.0004,0.0005]}
params = {'min_impurity_decrease':np.arange(0.0001,0.0005,0.0001)}
# n_jobs = -1 cpu의 모든 core 사용
gs = GridSearchCV(dt,params,n_jobs= -1)
gs.fit(train_data,train_label)
print()
```

```python
# 최적의 모델을 적용
dt = gs.best_estimator_
# 가장 좋은 파라미터,옵션값 출력
print(gs.best_params_)
print('test 평균정확도 : ',gs.cv_results_['mean_test_score'])
```

```python
dt.fit(train_data,train_label)
print("train정확도 : ",dt.score(train_data,train_label))
print("test정확도 : ",dt.score(test_data,test_label))
```
