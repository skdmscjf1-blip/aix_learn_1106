```python
#그래프관련 import
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
# matplotlib.rcParams['font.family'] = 'AppleGothic'
matplotlib.rcParams['axes.unicode_minus']=False

import pandas as pd
import numpy as np
from sklearn import svm,metrics
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier,KNeighborsRegressor
from sklearn.linear_model import LinearRegression,Ridge,Lasso,LogisticRegression
from sklearn.preprocessing import StandardScaler,PolynomialFeatures,OneHotEncoder
from sklearn.metrics import mean_absolute_error
from scipy.special import expit,softmax # expit : 이진분류함수 , softmax : 다중분류 0-1 숫자로 변환
from sklearn.linear_model import SGDClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import plot_tree
from sklearn.model_selection import cross_validate
from sklearn.model_selection import StratifiedKFold


```

```python
# 데이터 가져오기
wine = pd.read_csv('data/wine.csv')
wine.head(2)
data = wine[wine.columns[:-1]].values
label = wine['class'].values
# 데이터 분리
train_data,test_data,train_label,test_label = train_test_split(data,label,test_size=0.2,random_state=42)
```

```python
# default : test score만 출력 , return_train_score = True :  train score , test score 모두 출력
# 분류 : StratifiedKFold()
# 예측 : KFold()
dt = DecisionTreeClassifier(max_depth=4,random_state=42)
# n_splits=10 : 분할개수
# shuffle = True : 랜덤으로 섞어서 테스트 
splitter1 = StratifiedKFold(n_splits=10,shuffle=True,random_state=42) 
score = cross_validate(dt,train_data,train_label,cv=splitter1) #교차검증 5등분
score
```

```python
print('test 평균정확도',np.mean(score['test_score']))
```
