```python
#그래프관련 import
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
# matplotlib.rcParams['font.family'] = 'AppleGothic'
matplotlib.rcParams['axes.unicode_minus']=False

import pandas as pd
import numpy as np
from sklearn import svm,metrics
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier,KNeighborsRegressor
from sklearn.linear_model import LinearRegression,Ridge,Lasso,LogisticRegression
from sklearn.preprocessing import StandardScaler,PolynomialFeatures,OneHotEncoder
from sklearn.metrics import mean_absolute_error
from scipy.special import expit,softmax # expit : 이진분류함수 , softmax : 다중분류 0-1 숫자로 변환
from sklearn.linear_model import SGDClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import plot_tree
```

```python
# 데이터 가져오기
wine = pd.read_csv('data/wine.csv')
wine.head(2)
data = wine[wine.columns[:-1]].values
label = wine['class'].values
# 데이터 분리
train_data,test_data,train_label,test_label = train_test_split(data,label,random_state=42)
```

```python
# 특성값을 랜덤으로 설정해서 Tree구조
dt = DecisionTreeClassifier(max_depth=3,random_state=42)
dt.fit(train_data,train_label)
print()
```

```python
wine.columns
```

```python
# 그래프 지원

plt.figure(figsize=(10,7))
plot_tree(dt,max_depth=1,filled=True,feature_names=['alcohol', 'sugar', 'pH'])
plt.show()
```

```python
# 각 특성중에 어떤특성이 분류에 가장 영향을 줬는지 알수 있음.
print(wine.columns)
print(dt.feature_importances_)
```

```python
print('train정확도 : ',dt.score(train_data,train_label))
print('test정확도 : ',dt.score(test_data,test_label))
```
