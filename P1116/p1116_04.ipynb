```python
#그래프관련 import
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
# matplotlib.rcParams['font.family'] = 'AppleGothic'
matplotlib.rcParams['axes.unicode_minus']=False

import pandas as pd
import numpy as np
from sklearn import svm,metrics
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier,KNeighborsRegressor
from sklearn.linear_model import LinearRegression,Ridge,Lasso,LogisticRegression
from sklearn.preprocessing import StandardScaler,PolynomialFeatures,OneHotEncoder
from sklearn.metrics import mean_absolute_error
from scipy.special import expit,softmax # expit : 이진분류함수 , softmax : 다중분류 0-1 숫자로 변환
from sklearn.linear_model import SGDClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import plot_tree
from sklearn.model_selection import cross_validate
from sklearn.model_selection import StratifiedKFold
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier
```

```python
df = pd.read_csv('data/iris.csv',index_col='caseno')
df.head()
```

```python
data = df.iloc[:,:-1].values
label = df['Species'].values
```

```python
train_data,test_data,train_label,test_label = train_test_split(data,label,test_size=0.2,random_state=42)
```

```python
rf = RandomForestClassifier(n_jobs=-1,n_estimators=100,random_state=42)
rf.fit(train_data,train_label)

print('train정확도 : ',rf.score(train_data,train_label))
print('test정확도 : ',rf.score(test_data,test_label))

```

```python
rf = RandomForestClassifier(n_jobs=-1,n_estimators=100,random_state=42)
scores = cross_validate(rf,train_data,train_label,return_train_score=True,n_jobs=-1)
print(scores['train_score'])
print(scores['test_score'])
print(np.mean(scores['train_score']))
print(np.mean(scores['test_score']))
```
